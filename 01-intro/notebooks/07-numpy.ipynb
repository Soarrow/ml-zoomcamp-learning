## Creating arrays


```python
# Creating arrays the easiest way is to use zeros it takes in the size of the arr you want land fills it with zeros
np.zeros(10)
```

```python
np.ones(10)
```

```python
# you specify the size and the number you want to fill the arr with in this case we're making 10 2.5s
np.full(10, 2.5)
```

```python
a = np.array([1,2,3,5,7,12]) # converts python list into numpy arr
a
```

```python
a[2]
```

```python
a[2] = 10
a
```

```python
np.arange(3,10) # it creates an arr using a range of numbers - excluding the end number
```

```python
np.linspace(0,100,11) # create an arr using the range in this case and spaces them evenly to the end using the third parameter
# This fu0nction is inclusive of the beginning and the end
```

## Randomly generated arrays


```python
np.random.rand(5,2) # generate a 2D array of 5 rows and 2 cols with random number between 0 and 1 from standard uniform distribution
```

```python
np.random.seed(2)
100 * np.random.rand(5,2) 
# we also have rand and randn - which gives us random normal i.e. draw numbers from the standard normal distribution
```

```python
# for random integers we have randint
np.random.seed(2)
np.random.randint(low=0, high=100, size=(5,2))
```
